## CSV merge and binary

```python
import pandas as pd

# Load the data
red_wine_path = "C:/Users/Blake/Downloads/wine+quality/winequality-red.csv"
white_wine_path = "C:/Users/Blake/Downloads/wine+quality/winequality-white.csv"

# would not work without delimiter specified
red_wine = pd.read_csv(red_wine_path, delimiter=';')
white_wine = pd.read_csv(white_wine_path, delimiter=';')

# Add a binary
red_wine['wine_type'] = 1  
white_wine['wine_type'] = 0  

# Concatate both datasets
wine_data = pd.concat([red_wine, white_wine])

# Save the new dataframe to a CSV file
output_path = "C:/Users/Blake/Downloads/wine_data.csv"
wine_data.to_csv(output_path, index=False)

print("Data has been updated and saved to", output_path)

```
