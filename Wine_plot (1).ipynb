```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

red_path = "C:/Users/Blake/Downloads/wine+quality/winequality-red.csv"
white_path = "C:/Users/Blake/Downloads/wine+quality/winequality-white.csv"

red_wine = pd.read_csv(red_path, delimiter=';')
white_wine = pd.read_csv(white_path, delimiter=';')

# again assign binary classifier
red_wine['type'] = 1
white_wine['type'] = 0

# merge
data = pd.concat([red_wine, white_wine])

# grid the plot
sns.set(style="whitegrid")

# generate plot
plt.figure(figsize=(8,6))
boxplot = sns.boxplot(x='type', y='quality', data=data, palette=["lightblue", "darkred"])

# Formatting and aesthetic stuff
plt.xticks([0, 1], ['White', 'Red'])
plt.title('Box Plot: Wine Quality of White and Red Wines', fontsize=16)
plt.xlabel('Wine Type', fontsize=14)
plt.ylabel('Quality Score', fontsize=14)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

red_path = "C:/Users/Blake/Downloads/wine+quality/winequality-red.csv"
white_path = "C:/Users/Blake/Downloads/wine+quality/winequality-white.csv"

red_wine = pd.read_csv(red_path, delimiter=';')
white_wine = pd.read_csv(white_path, delimiter=';')

# again assign binary classifier
red_wine['type'] = 1
white_wine['type'] = 0

# merge
data = pd.concat([red_wine, white_wine])

# convert numeric type to string label for cleaner legend
data['type_label'] = data['type'].map({0: 'White', 1: 'Red'})

# grid the plot
sns.set(style="whitegrid")

# generate plot
plt.figure(figsize=(8,6))
scatter = sns.scatterplot(
    x='alcohol',
    y='quality',
    hue='type_label',
    data=data,
    palette={'White': 'lightblue', 'Red': 'darkred'},
    alpha=0.6
)

# Formatting and aesthetic stuff
plt.title('Scatter Plot: Wine Quality vs Alcohol Content', fontsize=16)
plt.xlabel('Alcohol', fontsize=14)
plt.ylabel('Quality Score', fontsize=14)
plt.legend(title='Wine Type')

plt.show()
```
